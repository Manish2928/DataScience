# Handling Imbalanced Dataset


1.   Up Sampling
2.   Down Sampling



```python
import numpy as np
import pandas as pd

# Set the Randome Seed for reproducibility
np.random.seed(123)

# Create a data frame with 2 classes
n_samples = 1000
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

```python
## Create My Class with Imbalanced dataset
class_0 = pd.DataFrame({
    'feature-1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature-2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature-1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature-2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

```python
## Upsampling

df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample
df_minority_Upsampled = resample(df_minority, replace=True,#Sample with Replacment
                                 n_samples = len(df_majority),
                                 random_state=42)
```

```python
df_minority_Upsampled.shape
```

```python
df_minority_Upsampled.head()
```

```python
df_upsampled = pd.concat([df_majority,df_minority_Upsampled])
```

```python
df_upsampled['target'].value_counts()
```

# Down Sampling

```python
## Downsampling

df_minority = df[df['target']==1]
df_majority = df[df['target']==0]
```

```python
from sklearn.utils import resample
df_majority_downsampled = resample(df_majority, replace=False,
                                 n_samples = len(df_minority),
                                 random_state=42)
```

```python
df_majority_downsampled.shape
```

```python
df_downsampled = pd.concat([
    df_majority_downsampled,
    df_minority
])
```

```python
df_downsampled.target.value_counts()
```
