# **Label And Ordinal Encoding**


*   Convert categorical data into numbers, but the important difference is whether the categories have a natural order.




```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
## Create a simple dataframe
df = pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
df.head()
```

```python
lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
lbl_encoder.transform([['red']])
```

```python
lbl_encoder.transform([['blue']])
```

# **Ordinal Encoding**



*   Ordinal Encoding is used when categories have a real, meaningful order.



```python
## Ordinal Encoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
# create a sample dataframe with an ordinal  variable
df = pd.DataFrame({
    'size':['small','medium','large','medium','small','large']
})
```

```python
df
```

```python
## Create an instance of OrdinalEncoder then fit_transform
encoder = OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoder.fit_transform(df[['size']])
```

```python
new_df = pd.concat([df,pd.DataFrame(encoder.fit_transform(df[['size']]),columns=['size_encoded'])],axis=1)
```

```python
new_df
```

# **Target Guided Ordinal Encoding**


*   Target Guided Ordinal Encoding is a technique where we encode categories based on their relationship with the target variable.


*   It is a technique used to encode categorical variables based on their relationship with the target variable. This encoding technique is useful
when we have a categorical variable with a large number of unique categories, and we want to use this variable as a feature in our machine
learning model.

* In Target Guided Ordinal Encoding, we replace each category in the categorical variable with a numerical value based on the mean or median
of the target variable for that category. This creates a monotonic relationship between the categorical variable and the target variable, which
can improve the predictive power of our model.





```python
# create a sample dataframe with a categorical variable and target variable

df = pd.DataFrame({
    'city':['New York','London','Paris','Tokyo','New York','Paris'],
    'price':[200,150,300,250,150,320]
})
```

```python
df
```

```python
mean_price = df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded'] = df['city'].map(mean_price)
```

```python
df
```

```python
df[['price','city_encoded']]
```
