# **Handling Outliers Using Python**
5 Number summary and box ploat

```python
## Minimum , maximum , Q1,Q3, IQR
import numpy as np
```

```python
lst_marks = [45,34,23,65,34,76,39,44,53,87,67,89,82,75,47,78,61,82,88,53,66,71,38]
minimum,Q1,median,Q3,maximum = np.quantile(lst_marks,[0,0.25,0.50,0.75,1.0])
```

```python
minimum,Q1,median,Q3,maximum
```

```python
IQR = Q3-Q1
print(IQR)
```

```python
lower_fance = Q1-1.5*(IQR)
higher_fance = Q3+1.5*(IQR)
```

```python
lower_fance
```

```python
higher_fance
```
